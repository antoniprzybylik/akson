# Example 6: Exporting PID Controller PLC Code

This example demonstrates how to export a configured MIMO (Multi-Input Multi-Output) PID controller to PLC (Programmable Logic Controller) code. This enables deployment of Akson-designed controllers directly to industrial hardware.

## Importing Akson

```python
import akson
```

## PID Controller Configuration

The PID controller is configured with three channels:

```python
config = akson.PIDControllerConfiguration(
    n_inputs=2, n_outputs=2,
    channels=[
        akson.PIDChannel(output_idx=0, input_idx=0, K=1., Ti=10., Td=0.0001),
        akson.PIDChannel(output_idx=1, input_idx=1, K=2., Ti=20., Td=0.0002),
        akson.PIDChannel(output_idx=0, input_idx=1, K=3., Ti=30., Td=0.0003)
    ],
    t=0.015
)
```

## Exporting PLC code

The code for PLC deployment is generated:

```python
plc_code = config.export_plc_code(
    setpoint_registers={0: 'D100', 1: 'D110'},
    controller_input_registers={0: 'D200', 1: 'D210'},
    controller_output_registers={0: 'D300', 1: 'D310'}
)
print(plc_code)
```
